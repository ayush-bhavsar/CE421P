```python
# Step 1: Import required libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.preprocessing import StandardScaler

```

```python
# Step 2: Generate comprehensive salary dataset
# More realistic salary data with more samples
data = {
    'YearsExperience': [0, 1, 1, 2, 2, 2, 3, 3, 4, 4, 5, 5, 5, 6, 6, 7, 7, 8, 8, 9, 9, 10, 10, 11, 11, 12, 12, 13, 13, 14, 15],
    'Salary': [30000, 32000, 33000, 35000, 36000, 37000, 40000, 42000, 45000, 47000, 50000, 52000, 53000, 55000, 57000, 60000, 62000, 65000, 67000, 70000, 72000, 75000, 78000, 80000, 82000, 85000, 87000, 90000, 92000, 95000, 98000]
}
df = pd.DataFrame(data)
df.to_csv("salary_data.csv", index=False)  # Save to CSV file

#print("CSV file created successfully!")

# Step 3: Load dataset from CSV
dataset = pd.read_csv("salary_data.csv")
print("\nDataset:")
print(dataset)
print(f"\nDataset shape: {dataset.shape}")
print(f"Dataset Info:\n{dataset.describe()}")
```

```python
# Step 4: Preprocessing
## (a) Handle missing values
#dataset = dataset.fillna(dataset.mean())

## (b) Features (X) and target (y)
X = dataset[['YearsExperience']]
y = dataset['Salary']

## (c) Feature Scaling (important for regression sometimes)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Step 5: Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Step 6: Build Linear Regression model
model = LinearRegression()
model.fit(X_train, y_train)

# Step 7: Predictions
y_pred = model.predict(X_test)

# Step 8: Accuracy (R² score)
accuracy = r2_score(y_test, y_pred)

# values that the machine learning model calculates
#(or predicts) after learning patterns from the training data.

print("\nPredicted Salaries:", y_pred)
print("Actual Salaries:", list(y_test)) # actual values in dataset
print(f"\nModel Accuracy (R² Score): {accuracy:.2f}")
```
