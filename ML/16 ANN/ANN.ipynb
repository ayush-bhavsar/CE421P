```python
# Import libraries
import pandas as pd
import numpy as np

#Imports Sequential model from Keras.
#This is a linear stack of layers → easiest way to build ANN
from tensorflow.keras.models import Sequential

#Imports Dense layer.
#A dense layer means: Every neuron is connected to every neuron in the previous layer.
from tensorflow.keras.layers import Dense

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
# Step 1: Load dataset
data = pd.read_csv("./heart_disease_500.csv")
```

```python
# Step 2: Split features and target
X = data.drop("target", axis=1)
y = data["target"]
```

```python
# Step 3: Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python

# Step 4: Feature scaling
scaler = StandardScaler()

#Learns scaling parameters from training data:
#Mean
#Standard deviation
#Then transforms training data.
X_train = scaler.fit_transform(X_train)


X_test = scaler.transform(X_test)

```

```python
# Step 5: Create ANN model Initializes empty ANN model.
model = Sequential()

# Hidden layers

#First hidden layer:
#8 neurons
#input_dim=X.shape[1] → number of features
#relu activation:
#Converts negative values to 0
#Helps model learn non-linear patterns

model.add(Dense(8, input_dim=X.shape[1], activation='relu'))

#Second hidden layer:
#4 neurons
#Adds more learning capability

model.add(Dense(4, activation='relu'))

# Output layer



#Output layer:
#1 neuron (binary classification)
#sigmoid:
#Outputs value between 0 and 1
#Interpreted as probability

model.add(Dense(1, activation='sigmoid'))

```

```python
# Step 6: Compile model

#loss='binary_crossentropy'
#Measures error for binary classification
#optimizer='adam'
#Adjusts weights efficiently
#Combines advantages of multiple optimizers
#metrics=['accuracy']
#Tracks performance during training

model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

```

```python
# Step 7: Train model

#Trains the ANN:
#epochs=50 → model sees data 50 times
#(An epoch is one complete pass of the entire training dataset through a machine learning model.)
#verbose=1 → shows training progress
#During training: Weights (W) and bias (b) are updated repeatedly

model.fit(X_train, y_train, epochs=50, verbose=1)

```

```python
# Step 8: Evaluate model
loss, accuracy = model.evaluate(X_test, y_test)
print("Accuracy:", accuracy)
```

```python

# Step 9: Predict

#Predicts results for first 5 test samples.
predictions = model.predict(X_test[:5])
print("Predictions:", predictions)
```
