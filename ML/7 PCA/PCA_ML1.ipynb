In **Machine Learning**, data often comes in different ranges. Example:

* Age: 0–100
* Salary: 20,000–1,00,000
* Height: 150–200

If we directly feed such uneven data to ML models, some features may dominate others. To avoid this, we **scale** (resize) the data. Two common methods are:

---

### 1. **Min-Max Scaler (Normalization)**

* Formula:

  $$
  X_{scaled} = \frac{X - X_{min}}{X_{max} - X_{min}}
  $$
* It **shrinks all values between 0 and 1** (sometimes between -1 and 1, depending on setting).

👉 Example:
If **age = 50**, and in the dataset:

* min age = 0
* max age = 100

Then,

$$
\frac{50 - 0}{100 - 0} = 0.5
$$

So **50 becomes 0.5**.
Now all ages will lie between **0 and 1**.

✅ Best when data is bounded and we want to preserve the shape of distribution.
⚠️ Sensitive to outliers (a very large/small value can distort scaling).

---

### 2. **Standard Scaler (Z-score Normalization)**

* Formula:

  $$
  X_{scaled} = \frac{X - mean}{standard \, deviation}
  $$
* It **centers data around 0** with a standard deviation of **1**.

👉 Example:
Suppose salaries: \[20k, 30k, 40k, 50k, 60k]

* Mean = 40k
* Std Dev ≈ 14.14k

For salary = 50k:

$$
\frac{50000 - 40000}{14142} \approx 0.71
$$

So **50k becomes 0.71**.
Values below mean become negative, above mean become positive.

✅ Good when data follows a **normal distribution (bell curve)**.

⚠️ Not bounded (values can be any number, not just 0–1).

---



### Quick Difference

| Feature      | Min-Max Scaler            | Standard Scaler                   |
| ------------ | ------------------------- | --------------------------------- |
| Range        | 0 to 1 (or -1 to 1)       | No fixed range                    |
| Formula      | (X – min)/(max – min)     | (X – mean)/std                    |
| Use case     | When bounds are important | When data is normally distributed |
| Sensitive to | Outliers                  | Less sensitive                    |

---

👉 In short:

* **Min-Max Scaler** = squish everything between 0 and 1.
* **Standard Scaler** = shift data so mean = 0, spread = 1.


```python
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

# Sample Data
data = {'Math': [80, 70, 90], 'Science': [85, 75, 88], 'English': [78, 68, 85]}
df = pd.DataFrame(data)

# Step 1: Standardize the Data
scaler = StandardScaler()
data_scaled = scaler.fit_transform(df)

# Step 2: Apply PCA
pca = PCA(n_components=2)
principal_components = pca.fit_transform(data_scaled)

# Step 3: Visualize the Results
plt.scatter(principal_components[:, 0], principal_components[:, 1], color='blue')
plt.title('PCA: Student Performance')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.show()
```
